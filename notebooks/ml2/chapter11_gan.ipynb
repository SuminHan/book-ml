헷갈려하는 \\(D=0.5\\)) 지점에서 두 손실이 어떤 값을 갖는지
확인합니다.

## 1. 판별자/생성자 손실 (책 10.2절, 연습문제 예제 그대로)

```python
import math

def discriminator_loss(D_real, D_fake):
    return -(math.log(D_real) + math.log(1 - D_fake))

def generator_loss(D_fake):
    return -math.log(D_fake)

print(discriminator_loss(D_real=0.9, D_fake=0.1))  # 판별자가 잘 구별할수록 손실이 작음
print(generator_loss(D_fake=0.9))                  # 생성자가 잘 속일수록 손실이 작음
```

## 2. 내쉬 균형(\\(D=0.5\\))에서의 손실값

이론상 균형점은 \\(D^*(x) = p_{\text{data}}/(p_{\text{data}}+p_G)\\)이고,
\\(p_G = p_{\text{data}}\\)가 되면 \\(D^*(x)=0.5\\)입니다. 이 지점에서
판별자 손실이 어떻게 되는지 직접 계산해봅니다.

```python
def optimal_discriminator(p_data, p_G):
    return p_data / (p_data + p_G)

# p_G가 p_data에 점점 가까워질 때 D*가 0.5로 수렴하는 과정
for p_G in [0.9, 0.5, 0.2, 0.05, 0.0]:
    p_data = 1.0
    d_star = optimal_discriminator(p_data, p_G if p_G > 0 else 1e-9)
    print(f"p_G={p_G:>4}, p_data={p_data} -> D*(x) = {d_star:.3f}")

d_at_equilibrium = optimal_discriminator(1.0, 1.0)
assert abs(d_at_equilibrium - 0.5) < 1e-9
print(f"\np_G == p_data 일 때 D*(x) = {d_at_equilibrium}")
print(f"균형점에서의 판별자 손실 = {discriminator_loss(0.5, 0.5):.4f}  (= 2 log 2, 정보이론적 최댓값)")
print(f"균형점에서의 생성자 손실   = {generator_loss(0.5):.4f}")
```
